**List all tables**

```python
tables = pd.read_sql_query("SELECT name, type FROM sqlite_master WHERE type IN ('table','view') ORDER BY name;", connection)
tables
```

**Exploring each tables**

```python
#Display some rows from the table
def preview(table, n=5):
    return pd.read_sql_query(f"SELECT * FROM {table} LIMIT {n};", connection)

#Count the total rows in the table
def row_count(table):
    return pd.read_sql_query(f"SELECT COUNT(*) AS rows FROM {table};", connection)

#Count the NULL values per column
def column_nulls(table):
    df = pd.read_sql_query(f"SELECT * FROM {table} LIMIT 50000;", connection)
    # small sample is enough to estimate
    return df.isnull().sum()

#Count distinct valules in each columns (up to 50000 samples)
def distinct_counts(table):
    df = pd.read_sql_query(f"SELECT * FROM {table} LIMIT 50000;", connection)
    return df.nunique()

#Show the summary of basic statistics
def numeric_summary(table):
    df = pd.read_sql_query(f"SELECT * FROM {table} LIMIT 50000;", connection)
    numeric = df.select_dtypes(include="number")
    return numeric.describe()
```

**1. Titles table**

```python
table = "titles"

#Display some rows from the Titles table
display(preview(table, 1000))

#Count the total rows in the table
display(row_count(table))

#Count the NULL values per column
display(column_nulls(table))

#Count distinct valules in each columns (up to 50000 samples)
display(distinct_counts(table))

#Show the summary of basic statistics
display(numeric_summary(table))
```
